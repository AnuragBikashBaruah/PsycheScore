## Encoding  strategy used
##### label encoding introduces a fake mathematical order and hierarchy where none exists , and this might lead the model for biasing toward one cat eg Male or Female so we will use onehot for those and will use label encoding for the columns that show a order like Stress_Level in order to maintain the order of our final model

## Train Test Split

```python
num_cols.columns
```

```python
from sklearn.model_selection import train_test_split

# for skewness handling
skew_col = ['Study_Hours']  # log transform

other_numeric_cols= ['Age', 'Avg_Daily_Usage_Hours', 'Daily_Unlocks','Physical_Activity_Hours', 'Sleep_Hours_Per_Night'] # standard scaler

# ordinal encoding and not label encoding
ordinal_col = ['Stress_Level'] 

# one hot encoding
normal_cols= ['Gender','Academic_Level','Most_Used_Platform','Purpose_Of_Use','Grouped_country']

feature_cols= skew_col + other_numeric_cols + ordinal_col + normal_cols
X=df[feature_cols]
y=df['Mental_Health_Score']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

## Preprocessing using ColumnTransformer

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer

# for skewed features
skew_pipeline=Pipeline(steps=[
  ('log_transform', FunctionTransformer(np.log1p)),
  ('scale',StandardScaler())
])

# for numeric features
plain_numeric_pipeline=Pipeline(steps=[
  ('scale',StandardScaler())
])

# for ordinal features
ordinal_pipeline=Pipeline(steps=[
  ('encode',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])

# for normal features
normal_pipeline=Pipeline(steps=[
  ('encode',OneHotEncoder(handle_unknown='ignore'))  # null value or nan values are ignored 
])
```

```python
preprocessor = ColumnTransformer(transformers=[
  ("Skewed_Pipeline", skew_pipeline, skew_col),
  ("Plain_Numeric", plain_numeric_pipeline, other_numeric_cols),
  ("Ordinal", ordinal_pipeline, ordinal_col),
  ("Normal", normal_pipeline, normal_cols)
])

preprocessor
```

## BaseLine: Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

lr_pipeline=Pipeline(steps=[
  ('preprocessor', preprocessor),
  ('regressor', LinearRegression())
])

lr_pipeline.fit(X_train,y_train) 
```

```python
lr_pred = lr_pipeline.predict(X_test)
lr_pred_train = lr_pipeline.predict(X_train)

lr_r2_testing = r2_score(y_test,lr_pred)
lr_r2_training = r2_score(y_train, lr_pred_train)
lr_mae = mean_absolute_error(y_test,lr_pred)

print(f"Accuracy of Training {lr_r2_training}")
print(f"Accuracy of Testing {lr_r2_testing}")
print(f"Mean Absolute error is {lr_mae}")
```

## Random Forest (default)

```python
from sklearn.ensemble import RandomForestRegressor

rf_pipeline=Pipeline(steps=[
  ('preprocessor', preprocessor),
  ('random forest', RandomForestRegressor(random_state=42))
])

rf_pipeline.fit(X_train, y_train)
```

```python
rf_pred = rf_pipeline.predict(X_test)
rf_pred_train = rf_pipeline.predict(X_train)

rf_r2_testing = r2_score(y_test, rf_pred)
rf_r2_training = r2_score(y_train, rf_pred_train)
rf_mae = mean_absolute_error(y_test, rf_pred)

print(f"Accuracy of Random Forest Training {rf_r2_training}")
print(f"Accuracy of Random Forest Testing {rf_r2_testing}")
print(f"Mean Absolute Error of Random Forest {rf_mae}")
```

## HyperParameter Tuning of Random Forest

```python
param_grid = {
  'random forest__n_estimators'       : [100,200,300],
  'random forest__max_depth'         : [5,10,15],
  'random forest__min_samples_split' : [2,5,10],
  'random forest__min_samples_leaf'  : [1,2,4]
}
```

```python
from sklearn.model_selection import RandomizedSearchCV

random_search = RandomizedSearchCV(
  estimator= rf_pipeline,
  param_distributions= param_grid,
  n_iter= 15,
  cv= 5,
  scoring= 'r2',
  random_state= 42,
  n_jobs= -1
)

random_search.fit(X_train,y_train)
```

```python
random_search.best_params_
```

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_pipeline
```

```python
rf_best_pred = rf_best_pipeline.predict(X_test)
rf_best_pred_train = rf_best_pipeline.predict(X_train)

rf_best_r2_testing = r2_score(y_test, rf_best_pred)
rf_best_r2_training = r2_score(y_train, rf_best_pred_train)
rf_best_mae = mean_absolute_error(y_test, rf_best_pred)

print(f"Accuracy of Random Forest Training after hyperparameter tuning {rf_best_r2_training}")
print(f"Accuracy of Random Forest Testing after hyperparameter tuning {rf_best_r2_testing}")
print(f"Mean Absolute Error of Random Forest after hyperparameter tuning {rf_best_mae}")
```

## Evaluation

```python
from sklearn.metrics import mean_squared_error

# Calculate RMSE for Linear Regression
lr_rmse = np.sqrt(mean_squared_error(y_test, lr_pred))

# Calculate RMSE for default Random Forest
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))

# Calculate RMSE for tuned Random Forest
rf_tuned_preds          = random_search.best_estimator_.predict(X_test)
rf_tuned_rmse           = np.sqrt(mean_squared_error(y_test, rf_best_pred))
rf_tuned_training_preds = random_search.best_estimator_.predict(X_train)
r2_tuned_training       = r2_score(y_train, rf_best_pred_train)
rf_tuned_mae            = mean_absolute_error(y_test, rf_best_pred)
rf_tuned_r2             = r2_score(y_test, rf_best_pred)


# Create a DataFrame to consolidate results
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest (default)', 'Random Forest (tuned)'],
    'R2': [lr_r2_testing, rf_r2_testing, rf_tuned_r2],
    'Training R2': [lr_r2_training, rf_r2_training, r2_tuned_training],
    'MAE': [lr_mae, rf_mae, rf_tuned_mae],
    'RMSE': [lr_rmse, rf_rmse, rf_tuned_rmse]
})

print(results)

# i will be using Random Forest (default)
```

## Saving the model
###### We save the entire pipeline — preprocessing and model together — as a single file, not just the raw model. Why: if we only saved the Random Forest object, our FastAPI backend would need to manually recreate every encoding and scaling step by hand at prediction time, which is fragile and easy to get wrong. Saving the full pipeline means the backend just loads one file and calls .predict() on raw input — the pipeline handles everything internally, exactly the same way it did during training.

```python
import joblib

joblib.dump(rf_pipeline, "../models/Mental_Health_Model.pkl")
```
